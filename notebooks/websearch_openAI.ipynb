```python
from agents import Agent, Runner, trace, WebSearchTool
from openai.types.responses import ResponseTextDeltaEvent
```

```python
nutrition_agent = Agent(name="Nutrition Assistant", 
instructions=""" you are a helpful assistant giving out nutrition advice. you give concise answers.
""", tools=[WebSearchTool()],
)
```

```python
with trace("Simple Nutrition Agent"):
    result = await Runner.run(nutrition_agent, "How healthy are cheese burgers?")

print(result)
```

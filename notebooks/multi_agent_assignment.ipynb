Agent Creation steps below

```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent

nutrition_agent = Agent(
name="Breakfast Advisor Agent",
instructions="""
1) Call the Agent Breakfast Planner to get the recommended breakfast items.
2) Call the Calorie Counter Agent for listing the calories for the breakfast items and the calories for each item.
3) Hand Off the process to Breakfast price Checkder Agent to break down the 
break fast item, ingredients with calories along with the price of each item and total.
 """
)

```

Invoke the Agent or Call the Agent

```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent

with trace("Simple Nutrition Agent"):
    result = await Runner.run(nutrition_agent, "How healthy is the english breakfast?")

print(result)
```
